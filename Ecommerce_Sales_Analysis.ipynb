```python
import matplotlib.pyplot as plt
```

```python
df['payment_method'].value_counts().plot(kind = 'bar', figsize = (8, 5))
plt.title("Payment Method Distribution")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.xticks(rotation = 45)
plt.show()
```

```python
df['delivery_status'].value_counts().plot(kind = 'bar', figsize = (8, 5))
plt.title("Delivery Status")
plt.xlabel("Status")
plt.ylabel("Count")
plt.xticks(rotation = 45)
plt.show()
```

```python
df['is_returned'].value_counts().plot(kind = 'pie', autopct = '%1.1f%%')
plt.title("Returned Products")
plt.ylabel("")
plt.show()
```

```python
df['category'].value_counts().plot(kind = 'bar', figsize = (10, 5))
plt.title("Top Product Categories")
plt.xticks(rotation = 45)
plt.show()
```

```python
df['brand'].value_counts().head(10).plot(kind = 'bar', figsize = (10, 5))
plt.title("Top 10 Brands")
plt.xticks(rotation = 45)
plt.show()
```

```python
df['rating'].plot(kind = 'hist', bins = 20, figsize = (8, 5))
plt.title("Ratings Distribution")
plt.xlabel("Rating")
plt.show()
```

```python
df['category'].value_counts().plot(kind = 'bar', figsize = (10, 5))
plt.title("Top Product Categories")
plt.xticks(rotation = 45)
plt.show()
```

```python
df.groupby('category')['rating'].mean().plot(kind = 'bar', figsize = (10, 5))
plt.title("Average Rating by Category")
plt.xticks(rotation = 45)
plt.show()
```

```python
df['payment_method'].value_counts().plot(kind = 'pie', autopct = '%1.1f%%')
plt.title("Payment Methods")
plt.show()
```

```python
df['delivery_status'].value_counts().plot(kind = 'bar', figsize = (8, 5))
plt.title("Delivery Status")
plt.xticks(rotation = 45)
plt.show()
```
