```python
import pandas as pd

file_path = '/content/drive/MyDrive/amazon_ecommerce_1M.csv'

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print(df.head())
```

```python
print(df['category'].value_counts())
```

```python
import matplotlib.pyplot as plt

df['category'].value_counts().plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Products by Category')
plt.xlabel('Category')
plt.ylabel('Number of Products')
plt.xticks(rotation=45)
plt.show()
```

```python
category_counts = df['category'].value_counts()

print("Category with most products:")
print(category_counts.idxmax())

print("\nNumber of products:")
print(category_counts.max())

print("\nCategory-wise product count:")
print(category_counts)
```

```python
print(df.columns.tolist())
```

```python
top_products = df.nlargest(10, 'review_count')

print(top_products[['product_id', 'category', 'review_count']])

plt.figure(figsize=(12, 5))

plt.bar(
    top_products['product_id'].astype(str),
    top_products['review_count']
)

plt.title('Top 10 Most Reviewed Products')
plt.xlabel('Product ID')
plt.ylabel('Number of Reviews')
plt.xticks(rotation=45)
plt.show()
```

```python
print("Average Product Price:", df['price'].mean())

print("Highest Product Price:", df['price'].max())

print("Lowest Product Price:", df['price'].min())
```

```python
# Average discount
print("Average Discount:", df['discount'].mean())

# Highest discount
print("Highest Discount:", df['discount'].max())

# Lowest discount
print("Lowest Discount:", df['discount'].min())

# Top 10 products with highest discounts
top_discount = df.nlargest(10, 'discount')

display(top_discount[['product_id', 'category', 'price', 'discount', 'final_price']])
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df['discount'], bins=20)

plt.title('Distribution of Product Discounts')
plt.xlabel('Discount')
plt.ylabel('Number of Products')

plt.show()
```

```python
# Count orders by delivery status
delivery_counts = df['delivery_status'].value_counts()

print("Delivery Status:")
print(delivery_counts)
```

```python
plt.figure(figsize=(8, 5))

delivery_counts.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Order Delivery Status Distribution')
plt.ylabel('')
plt.show()
```

```python
payment_counts = df['payment_method'].value_counts()

print("Payment Method Analysis:")
print(payment_counts)
```

```python
plt.figure(figsize=(9, 5))

payment_counts.plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title('Popular Payment Methods')
plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)

plt.show()
```

```python
print("Average Rating:", df['rating'].mean())

print("Highest Rating:", df['rating'].max())

print("Lowest Rating:", df['rating'].min())

# Top 10 highest-rated products
top_rated = df.nlargest(10, 'rating')

display(top_rated[['product_id', 'category', 'rating']])
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['rating'].dropna(), bins=10)

plt.title('Product Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Number of Products')

plt.show()
```

```python
return_counts = df['is_returned'].value_counts()

print("Product Return Analysis:")
print(return_counts)
```

```python
return_counts.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7, 7)
)

plt.title('Product Return Distribution')
plt.ylabel('')
plt.show()
```

```python
print("Average Shipping Time:",
      df['shipping_time_days'].mean())

print("Maximum Shipping Time:",
      df['shipping_time_days'].max())

print("Minimum Shipping Time:",
      df['shipping_time_days'].min())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['shipping_time_days'].dropna(), bins=15)

plt.title('Shipping Time Distribution')
plt.xlabel('Shipping Time (Days)')
plt.ylabel('Number of Orders')

plt.show()
```

```python
df['purchase_date'] = pd.to_datetime(
    df['purchase_date'],
    errors='coerce'
)

monthly_purchases = (
    df.dropna(subset=['purchase_date'])
      .set_index('purchase_date')
      .resample('ME')['product_id']
      .count()
)

print(monthly_purchases)
```

```python
plt.figure(figsize=(12, 5))

monthly_purchases.plot(kind='line', marker='o')

plt.title('Monthly Purchase Trends')
plt.xlabel('Month')
plt.ylabel('Number of Purchases')

plt.grid(True)
plt.show()
```

```python
seller_analysis = df.groupby('seller_id').agg(
    total_products=('product_id', 'count'),
    average_rating=('seller_rating', 'mean')
)

seller_analysis = seller_analysis.sort_values(
    by='total_products',
    ascending=False
)

print("Top 10 Sellers:")
display(seller_analysis.head(10))
```

```python
seller_analysis.head(10)['total_products'].plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Top 10 Sellers by Product Records')
plt.xlabel('Seller ID')
plt.ylabel('Number of Records')
plt.xticks(rotation=45)
plt.show()
```

```python
subcategory_counts = df['subcategory'].value_counts()

print("Top 10 Subcategories:")
print(subcategory_counts.head(10))
```

```python
subcategory_counts.head(10).plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Top 10 Product Subcategories')
plt.xlabel('Subcategory')
plt.ylabel('Number of Products')
plt.xticks(rotation=45)
plt.show()
```

```python
category_price = df.groupby('category')['final_price'].mean()

print("Average Final Price by Category:")
print(category_price)
```

```python
category_price.sort_values(ascending=False).plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Average Final Price by Category')
plt.xlabel('Category')
plt.ylabel('Average Final Price')
plt.xticks(rotation=45)
plt.show()
```

```python
total_sales = df['final_price'].sum()

average_sales = df['final_price'].mean()

print("Total Sales Value:", total_sales)
print("Average Purchase Value:", average_sales)
print("Total Transactions:", len(df))
```

```python
print("===== AMAZON ECOMMERCE ANALYSIS =====")

print("Total Records:", len(df))

print("Total Categories:", df['category'].nunique())

print("Most Popular Category:",
      df['category'].mode()[0])

print("Most Used Payment Method:",
      df['payment_method'].mode()[0])

print("Average Product Rating:",
      round(df['rating'].mean(), 2))

print("Average Shipping Time:",
      round(df['shipping_time_days'].mean(), 2), "days")

print("Total Sales Value:",
      round(df['final_price'].sum(), 2))

print("Total Returned Orders:",
      df['is_returned'].sum())
```
